## 1. Concepto de sentencia iterativa (criterio a)

Una **sentencia iterativa** (o **bucle**) **repite** un bloque de instrucciones. Cada repetición se llama **iteración**.

¿Por qué las necesitamos? Imagina imprimir los números del 1 al 1000 sin ellas.

````python
# Sin bucle (inviable a gran escala)
print(1)
print(2)
print(3)
# ... 997 líneas más

# Con bucle
for i in range(1, 6):
    print(i)
````

### Elementos de un bucle

1. **Inicialización**: valor inicial de las variables de control.
2. **Condición**: decide si se hace otra iteración.
3. **Cuerpo**: instrucciones que se repiten.
4. **Actualización**: modifica las variables para que el bucle termine.

```mermaid
flowchart TD
    A([Inicio]) --> B[Inicializar]
    B --> C{¿Condición?}
    C -- Sí --> D[Cuerpo del bucle]
    D --> E[Actualizar]
    E --> C
    C -- No --> F([Fin])
```

Python tiene dos bucles:

| Bucle | Cuándo |
|---|---|
| `for` | Se recorre una secuencia/colección o se conoce el número de repeticiones (**iteración definida**) |
| `while` | Se repite mientras se cumpla una condición (**iteración indefinida**) |

````python
# Los 4 elementos en un while
contador = 1                      # 1. inicialización
while contador <= 3:              # 2. condición
    print("Iteración", contador)  # 3. cuerpo
    contador += 1                 # 4. actualización
````

## 2. Estructuras condicionales vs. iterativas (criterio b)

| | Condicional (`if`) | Iterativa (`while`/`for`) |
|---|---|---|
| Objetivo | **Elegir** si/qué bloque se ejecuta | **Repetir** un bloque |
| Veces que se ejecuta el bloque | 0 o 1 | 0, 1 o muchas |
| Tras ejecutar el bloque | Continúa después del `if` | **Vuelve** a evaluar la condición |
| Riesgo típico | Orden/límites incorrectos | Bucle infinito |

````python
# Misma condición, distinto comportamiento
n = 3

print("if:")
if n > 0:
    print("  n =", n)
    n -= 1

print("while:")
n = 3
while n > 0:
    print("  n =", n)
    n -= 1
````

## 3. Verificación del funcionamiento de los bucles (criterio c)

Verificar un bucle es comprobar que:
- **Empieza** cuando debe (inicialización correcta).
- **Se repite** el número de veces esperado.
- **Termina** (la condición acaba siendo falsa).
- **Produce** el resultado correcto, incluidos los **casos límite** (0 iteraciones, 1 iteración, lista vacía...).

Técnicas: *prints* de depuración, **traza** manual, contadores de iteraciones, `assert` y probar valores límite.

````python
# Técnica 1: imprimir el estado en cada iteración
total = 0
for i in range(1, 5):
    total += i
    print(f"i={i}  total={total}")
````

````python
# Técnica 2: contar iteraciones y comprobar con assert
iteraciones = 0
for _ in range(10):
    iteraciones += 1
assert iteraciones == 10, "El bucle debería ejecutarse 10 veces"
print("Número de iteraciones correcto")
````

````python
# Técnica 3: casos límite
def suma(lista):
    total = 0
    for x in lista:
        total += x
    return total

assert suma([]) == 0                # lista vacía: 0 iteraciones
assert suma([5]) == 5               # 1 elemento
assert suma([1, 2, 3, 4]) == 10     # caso general
print("Todos los casos correctos ✔")
````

### 3.1 Errores típicos que hay que detectar

1. **Bucle infinito:** la condición nunca pasa a falsa.
2. **Error de límite (off-by-one):** una iteración de más o de menos.
3. **Variable no actualizada** o inicializada dentro del bucle (se reinicia cada vuelta).
4. **Modificar la lista mientras se recorre.**

````python
# Off-by-one: range(1, 5) NO incluye el 5
print(list(range(1, 5)))     # [1, 2, 3, 4]
print(list(range(1, 6)))     # [1, 2, 3, 4, 5]

# Variable inicializada dentro del bucle (error lógico)
for i in range(3):
    acumulado = 0            # ¡se reinicia en cada vuelta!
    acumulado += i
print("acumulado (incorrecto):", acumulado)

acumulado = 0
for i in range(3):
    acumulado += i
print("acumulado (correcto):", acumulado)
````

````python
# Modificar una lista mientras se recorre: resultado inesperado
numeros = [1, 2, 3, 4, 5, 6]
for n in numeros:
    if n % 2 == 0:
        numeros.remove(n)
print(numeros)          # [1, 3, 5] ... ¡pero con otras listas falla!

datos = [2, 4, 6, 8]
for n in datos:
    if n % 2 == 0:
        datos.remove(n)
print(datos)            # [4, 8] -> ¡no se eliminaron todos!

# Solución: recorrer una copia o construir una lista nueva
datos = [2, 4, 6, 8]
datos = [n for n in datos if n % 2 != 0]
print(datos)
````

````python
# Protección contra bucles infinitos durante las pruebas: límite de seguridad
n = 1
intentos = 0
while n != 0 and intentos < 5:      # n nunca llega a 0 -> el límite nos salva
    n += 1
    intentos += 1
print("Salimos por límite de seguridad. n =", n)
````

## 4. Bucles según las necesidades del problema (criterio d)

Analiza el problema y elige:

| Si necesitas... | Usa |
|---|---|
| Recorrer **todos** los elementos de una colección | `for elemento in coleccion` |
| Repetir **n** veces | `for _ in range(n)` |
| Recorrer con **índice** | `for i, x in enumerate(lista)` |
| Recorrer **varias** colecciones a la vez | `for a, b in zip(l1, l2)` |
| Repetir hasta que ocurra un evento (entrada válida, fin de datos) | `while` |
| Salir antes de tiempo | `break` |
| Saltar elementos | `continue` |

````python
# Caso 1: recorrer todos los elementos -> for
precios = [10, 20, 30]
print("Total:", sum(precios))
for p in precios:
    print("Precio:", p)
````

````python
# Caso 2: no se sabe cuántas veces -> while (entradas simuladas)
entradas = iter(["abc", "123", "1234"])     # simulan lo que escribiría el usuario
pin = next(entradas)
while pin != "1234":
    print(f"PIN '{pin}' incorrecto")
    pin = next(entradas)
print("Acceso concedido")
````

````python
# Caso 3: buscar y salir en cuanto se encuentre -> for + break
usuarios = ["ana", "luis", "marta", "pedro"]
buscado = "marta"

for i, usuario in enumerate(usuarios):
    if usuario == buscado:
        print(f"Encontrado en la posición {i}")
        break
else:                                   # se ejecuta si el bucle NO terminó con break
    print("No encontrado")
````

````python
# Caso 4: dos colecciones en paralelo -> zip
nombres = ["Ana", "Luis", "Marta"]
notas = [8.5, 4.0, 6.7]
for nombre, nota in zip(nombres, notas):
    estado = "aprobado" if nota >= 5 else "suspenso"
    print(f"{nombre}: {nota} ({estado})")
````

## 5. El bucle `for` (criterio e)

```python
for variable in iterable:
    cuerpo
```

Un **iterable** es cualquier objeto que se pueda recorrer: listas, tuplas, cadenas, diccionarios, conjuntos, `range()`, ficheros...

### 5.1 `range()`

| Forma | Resultado |
|---|---|
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(2, 6)` | 2, 3, 4, 5 |
| `range(0, 10, 3)` | 0, 3, 6, 9 |
| `range(5, 0, -1)` | 5, 4, 3, 2, 1 |

````python
print(list(range(5)))
print(list(range(2, 6)))
print(list(range(0, 10, 3)))
print(list(range(5, 0, -1)))
````

````python
# 5.2 Recorrer distintos iterables
for letra in "Hola":
    print(letra, end=" ")
print()

for elemento in (10, 20, 30):          # tupla
    print(elemento, end=" ")
print()

edades = {"Ana": 25, "Luis": 31}
for clave in edades:                    # claves
    print(clave, end=" ")
print()
for clave, valor in edades.items():     # claves y valores
    print(f"{clave} tiene {valor} años")
````

````python
# 5.3 enumerate y zip
frutas = ["manzana", "pera", "uva"]
for indice, fruta in enumerate(frutas, start=1):
    print(indice, fruta)
````

````python
# 5.4 break, continue y else en for
for n in range(1, 11):
    if n % 3 == 0:
        continue        # salta múltiplos de 3
    if n > 8:
        break           # termina el bucle
    print(n, end=" ")
print()

# for ... else: el else se ejecuta si el bucle terminó SIN break
for n in range(2, 7):
    if 7 % n == 0:
        print("7 no es primo")
        break
else:
    print("7 es primo")
````

````python
# 5.5 Patrones habituales con for: acumulador, contador, máximo, filtrado
numeros = [4, 9, 2, 15, 8]

suma = 0
cuenta_mayores = 0
maximo = numeros[0]
pares = []

for n in numeros:
    suma += n                       # acumulador
    if n > 5:
        cuenta_mayores += 1         # contador
    if n > maximo:
        maximo = n                  # máximo
    if n % 2 == 0:
        pares.append(n)             # filtrado

print(suma, cuenta_mayores, maximo, pares)
````

### 5.6 Comprensión de listas (forma compacta de un `for`)

````python
cuadrados = [n ** 2 for n in range(1, 6)]
pares = [n for n in range(1, 11) if n % 2 == 0]
print(cuadrados)
print(pares)
````

## 6. El bucle `while` (criterio e)

```python
while condicion:
    cuerpo       # debe modificar algo que afecte a la condición
```

Se comprueba la condición **antes** de cada iteración: el cuerpo puede ejecutarse **0 veces**.

````python
# 6.1 Bucle controlado por contador
i = 1
while i <= 5:
    print(i, end=" ")
    i += 1
print()
````

````python
# 6.2 Bucle controlado por centinela
datos = iter([3, 8, 2, -1, 99])     # -1 es el valor centinela
suma = 0
dato = next(datos)
while dato != -1:
    suma += dato
    dato = next(datos)
print("Suma:", suma)
````

````python
# 6.3 Bucle "infinito" controlado con break (equivalente al do-while de otros lenguajes)
intentos_simulados = iter(["x", "y", "python"])
while True:
    clave = next(intentos_simulados)
    if clave == "python":
        print("Clave correcta")
        break
    print("Clave incorrecta, reintenta")
````

````python
# 6.4 Aplicación: algoritmo de Collatz (número de pasos hasta llegar a 1)
n = 27
pasos = 0
while n != 1:
    n = n // 2 if n % 2 == 0 else 3 * n + 1
    pasos += 1
print("Pasos:", pasos)
````

````python
# 6.5 Aplicación: cuántos años tarda en duplicarse un capital al 5 %
capital_inicial = 1000
capital = capital_inicial
anios = 0
while capital < 2 * capital_inicial:
    capital *= 1.05
    anios += 1
print(f"Se duplica en {anios} años: {capital:.2f} €")
````

````python
# 6.6 while con else (se ejecuta si no hubo break)
n = 0
while n < 3:
    n += 1
else:
    print("El bucle terminó normalmente")
````

### 6.7 `for` frente a `while`: ejemplo comparado

````python
# Sumar 1..100 con for
total_for = 0
for i in range(1, 101):
    total_for += i

# Sumar 1..100 con while
total_while = 0
i = 1
while i <= 100:
    total_while += i
    i += 1

print(total_for, total_while, total_for == total_while)
````

## 7. Anidamiento de estructuras (criterio f)

Se pueden **anidar** bucles dentro de bucles, y bucles con condicionales (en cualquier combinación). El bucle interior se ejecuta **completo** en cada iteración del exterior:

> iteraciones totales = iteraciones exteriores × iteraciones interiores

### 7.1 Bucles anidados

````python
# Tabla de multiplicar completa (1..5)
for i in range(1, 6):
    for j in range(1, 6):
        print(f"{i * j:3}", end="")
    print()          # salto de línea al terminar la fila
````

````python
# Patrones con asteriscos
filas = 4
for i in range(1, filas + 1):
    print("*" * i)

print()
for i in range(1, filas + 1):
    for j in range(i):
        print("*", end="")
    print()
````

````python
# Recorrer una matriz (lista de listas)
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
]

suma = 0
for fila in matriz:
    for valor in fila:
        suma += valor
print("Suma total:", suma)

# Diagonal principal
for i in range(len(matriz)):
    print(matriz[i][i], end=" ")
print()
````

### 7.2 Bucle con condicional y condicional con bucle

````python
# for + if: números primos entre 2 y 30
primos = []
for n in range(2, 31):
    es_primo = True
    for d in range(2, int(n ** 0.5) + 1):
        if n % d == 0:
            es_primo = False
            break               # break solo sale del bucle INTERIOR
    if es_primo:
        primos.append(n)
print(primos)
````

````python
# Salir de dos bucles a la vez: se usa una bandera o una función con return
matriz = [[1, 2], [3, 4], [5, 6]]
objetivo = 4
encontrado = False

for i, fila in enumerate(matriz):
    for j, valor in enumerate(fila):
        if valor == objetivo:
            print(f"Encontrado en ({i}, {j})")
            encontrado = True
            break
    if encontrado:
        break
````

````python
# while dentro de for: reintentos por cada usuario (datos simulados)
usuarios = ["ana", "luis"]
claves_simuladas = {"ana": iter(["mal", "ok"]), "luis": iter(["ok"])}

for usuario in usuarios:
    intentos = 0
    while True:
        clave = next(claves_simuladas[usuario])
        intentos += 1
        if clave == "ok":
            print(f"{usuario}: acceso tras {intentos} intento(s)")
            break
````

````python
# Número de iteraciones de bucles anidados
contador = 0
for i in range(4):
    for j in range(3):
        contador += 1
print(contador)      # 4 x 3 = 12
````

## 8. Problemas integradores (criterios d, e, f)

### Problema 1: Menú con repetición (entradas simuladas)

````python
opciones = iter(["1", "2", "9", "3"])      # simula las elecciones del usuario
saldo = 100

while True:
    print("\n1) Consultar  2) Ingresar 50  3) Salir")
    opcion = next(opciones)
    if opcion == "1":
        print("Saldo:", saldo)
    elif opcion == "2":
        saldo += 50
        print("Ingreso realizado")
    elif opcion == "3":
        print("Hasta pronto")
        break
    else:
        print("Opción no válida")
````

### Problema 2: Estadísticas de una lista

````python
datos = [12, 7, 9, 21, 7, 15, 9, 7]

frecuencias = {}
for d in datos:
    frecuencias[d] = frecuencias.get(d, 0) + 1

moda = None
maxima = 0
for valor, veces in frecuencias.items():
    if veces > maxima:
        moda, maxima = valor, veces

print("Frecuencias:", frecuencias)
print(f"Moda: {moda} ({maxima} veces)")
````

## 1. ¿Qué es el flujo de control? (criterio a)

Un programa es una lista de instrucciones. El **flujo de control** es el orden en el que el intérprete las ejecuta. Por defecto, Python ejecuta de **arriba abajo**, una a una. Las **estructuras de control** permiten **modificar** ese orden:

| Tipo | Qué hace | En Python | Ejemplo de la vida real |
|---|---|---|---|
| **Secuencial** | Ejecuta instrucciones una tras otra | (por defecto) | Receta de cocina: pasos en orden |
| **Selección (condicional)** | Elige un camino según una condición | `if`, `elif`, `else`, `match` | Un cajero comprueba si hay saldo suficiente |
| **Repetición (iterativa)** | Repite un bloque mientras se cumpla algo o para cada elemento | `for`, `while` | Procesar cada producto del carrito de la compra |

Además existen sentencias que alteran el flujo *dentro* de los bucles: `break`, `continue` y `pass`.

### Aplicaciones reales

- **Secuencial:** calcular el total de una factura (base + IVA).
- **Selección:** un semáforo, un control de acceso, un descuento según la edad.
- **Repetición:** enviar un correo a cada cliente de una lista, reintentar una contraseña hasta 3 veces.

````python
# Un programa SECUENCIAL: cada línea se ejecuta una tras otra
print("1. Inicio")
print("2. Proceso")
print("3. Fin")

# Si cambias el orden de las líneas, cambia el resultado: el orden importa.
````

Un **diagrama de flujo** es la representación gráfica de un algoritmo. Símbolos estándar:

## 3. Importancia de las condiciones (criterio c)

Las estructuras de **selección** y de **repetición** dependen de una **condición**: una expresión que Python evalúa como `True` o `False`. Si la condición es incorrecta, el programa se comporta mal **aunque la sintaxis sea perfecta**.

### 3.1 Operadores que forman condiciones

| Tipo | Operadores |
|---|---|
| Comparación | `==`  `!=`  `<`  `>`  `<=`  `>=` |
| Lógicos | `and`  `or`  `not` |
| Pertenencia | `in`  `not in` |
| Identidad | `is`  `is not` |

````python
edad = 20
tiene_carnet = True

print(edad >= 18)                       # True
print(edad >= 18 and tiene_carnet)      # True
print(edad < 18 or not tiene_carnet)    # False
print("a" in "casa")                    # True
print(3 not in [1, 2, 4])               # True
````

### 3.2 Valores "verdaderos" y "falsos" (*truthy* / *falsy*)

En una condición, Python considera **falsos**: `False`, `None`, `0`, `0.0`, `""` (cadena vacía), `[]`, `()`, `{}`, `set()`. Todo lo demás es **verdadero**.

````python
for valor in [0, 1, "", "hola", [], [0], None]:
    print(f"{valor!r:8} -> {bool(valor)}")
````

### 3.3 Errores típicos con las condiciones

1. Confundir `=` (asignación) con `==` (comparación).
2. Olvidar los paréntesis cuando se mezclan `and` / `or` (`and` tiene más precedencia que `or`).
3. Condiciones que **nunca** se cumplen o **siempre** se cumplen (en un `while` produce un bucle infinito).
4. Comparar con `is` en lugar de `==` (valores).
5. Errores en los límites (`<` frente a `<=`).

````python
# Precedencia: and se evalúa antes que or
a, b, c = True, False, False
print(a or b and c)      # True  -> a or (b and c)
print((a or b) and c)    # False -> los paréntesis cambian el resultado

# Error de límites: ¿entra con 18 años?
edad = 18
print(edad > 18)    # False -> el de 18 queda fuera (probablemente un error)
print(edad >= 18)   # True
````

## 4. Importancia de los sangrados (criterio d)

A diferencia de otros lenguajes que usan llaves `{ }`, **Python delimita los bloques mediante el sangrado** (indentación). Todo lo que está indentado bajo una línea terminada en `:` pertenece a ese bloque.

**Reglas:**
- Usa **4 espacios** por nivel (estándar PEP 8). No mezcles tabuladores y espacios.
- Todas las líneas de un mismo bloque deben tener **el mismo** sangrado.
- Una línea sin sangrar marca el **fin** del bloque.

````python
edad = 20

if edad >= 18:
    print("Eres mayor de edad")     # dentro del if
    print("Puedes votar")           # dentro del if
print("Esta línea se ejecuta siempre")   # fuera del if
````

### 4.1 El sangrado cambia el significado del programa

Las dos versiones siguientes solo se diferencian en un sangrado, pero producen resultados distintos.

````python
print("Versión A: print dentro del bucle")
for i in range(3):
    print("Vuelta", i)
    print("Fin de la vuelta")

print()
print("Versión B: segundo print fuera del bucle")
for i in range(3):
    print("Vuelta", i)
print("Fin de la vuelta")
````

### 4.2 Errores de sangrado

Python los detecta como `IndentationError`. Los ejecutamos con `exec` dentro de un `try` para ver el mensaje sin romper el cuaderno.

````python
codigos_erroneos = {
    "Falta sangrado tras ':'": "if True:\nprint('hola')",
    "Sangrado inconsistente": "if True:\n    print('a')\n  print('b')",
    "Sangrado inesperado": "x = 1\n    y = 2",
}

for descripcion, codigo in codigos_erroneos.items():
    try:
        exec(codigo)
    except (IndentationError, SyntaxError) as e:
        print(f"{descripcion:28} -> {type(e).__name__}: {e.msg}")
````

## 5. Control secuencial (criterio e)

En un **bloque secuencial** las instrucciones se ejecutan en el orden en el que están escritas. Estructura típica de un programa: **entrada → proceso → salida**.

````python
# Ejemplo: factura con IVA
precio_base = 80.0      # entrada (en un programa real: float(input(...)))
iva = 0.21

importe_iva = precio_base * iva      # proceso
total = precio_base + importe_iva

print(f"Base:  {precio_base:8.2f} €")  # salida
print(f"IVA:   {importe_iva:8.2f} €")
print(f"Total: {total:8.2f} €")
````

````python
# Ejemplo: conversión de temperaturas
celsius = 25
fahrenheit = celsius * 9 / 5 + 32
kelvin = celsius + 273.15
print(f"{celsius} ºC = {fahrenheit} ºF = {kelvin} K")
````

````python
# Ejemplo: el orden importa (intercambio de dos variables)
a, b = 3, 7
temporal = a
a = b
b = temporal
print(a, b)          # 7 3

# Python permite hacerlo en una sola instrucción
a, b = b, a
print(a, b)          # 3 7
````

### ✏️ Actividad 5
Escribe un bloque secuencial que, dados los lados `base` y `altura` de un triángulo, calcule y muestre su **área** y su **perímetro** (suponiendo que es rectángulo: la hipotenusa es `(base**2 + altura**2) ** 0.5`).

````python
base = 3
altura = 4

def calcular_triangulo(b, h):
    # Cálculo correcto de la hipotenusa agrupando los términos con paréntesis
    hipotenusa = (b**2 + h**2)**0.5
    # Cálculo del área y el perímetro
    area = (b * h) / 2
    perimetro = b + h + hipotenusa
    return area, perimetro

area_calculada, perimetro_calculado = calcular_triangulo(base, altura)

# Mostrar los resultados de forma clara
print(f"Para un triángulo con base {base} y altura {altura}:")
print(f"Área: {area_calculada}")
print(f"Perímetro: {perimetro_calculado}")
````

## 6. Control de selección (criterio f)

La **selección** permite ejecutar un bloque **solo si** se cumple una condición.

```python
if condicion1:
    ...
elif condicion2:
    ...
else:
    ...
```

- `if`: obligatorio. 
- `elif`: opcional, tantos como se necesiten.
- `else`: opcional, siempre al final; se ejecuta si no se cumplió ninguna condición anterior.
- En una cadena `if/elif/else` **solo se ejecuta un bloque**.

````python
# Selección simple (if)
temperatura = 35
if temperatura > 30:
    print("Hace calor 🥵")

# Selección doble (if / else)
saldo, importe = 50, 80
if saldo >= importe:
    print("Pago aceptado")
else:
    print("Saldo insuficiente")
````

````python
# Selección múltiple (if / elif / else)
nota = 7.2

if nota < 5:
    calificacion = "Suspenso"
elif nota < 6:
    calificacion = "Suficiente"
elif nota < 7:
    calificacion = "Bien"
elif nota < 9:
    calificacion = "Notable"
else:
    calificacion = "Sobresaliente"

print(f"Nota {nota}: {calificacion}")
````

````python
# Aplicación real: precio de entrada de cine según la edad
edad = 70

if edad < 4:
    precio = 0
elif edad < 18:
    precio = 5
elif edad < 65:
    precio = 8
else:
    precio = 6

print(f"Edad {edad}: {precio} €")
````

## 7. Control de repetición (criterio g)

La **repetición** ejecuta un bloque varias veces.

| Bucle | Se usa cuando... | Sintaxis |
|---|---|---|
| `for` | Se sabe **cuántas veces** repetir o hay una **colección** que recorrer | `for elemento in coleccion:` |
| `while` | Se repite **mientras** se cumpla una condición (no se sabe cuántas veces) | `while condicion:` |

````python
# for: recorrer una secuencia
for numero in range(1, 6):          # 1, 2, 3, 4, 5
    print("Número:", numero)

frutas = ["manzana", "pera", "uva"]
for fruta in frutas:
    print("Fruta:", fruta)
````

````python
# while: repetir mientras la condición sea verdadera
contador = 3
while contador > 0:
    print("Cuenta atrás:", contador)
    contador -= 1       # ¡Imprescindible! Si no, bucle infinito
print("¡Despegue! 🚀")
````

````python
# break y continue
print("break: salir del bucle")
for n in range(1, 10):
    if n == 4:
        break
    print(n)

print("continue: saltar a la siguiente vuelta")
for n in range(1, 6):
    if n % 2 == 0:
        continue
    print(n)
````

````python
# Aplicación real: suma de los precios de un carrito
carrito = [12.5, 3.99, 8.0, 15.25]
total = 0
for precio in carrito:
    total += precio
print(f"Total del carrito: {total:.2f} €")
````

## 8. Combinación de estructuras: aplicación real

Los programas reales **combinan** las tres estructuras. Ejemplo: simulación de un cajero automático (con las entradas simuladas para poder ejecutar la celda).

````python
saldo = 300
operaciones = [("retirar", 100), ("retirar", 250), ("ingresar", 50), ("retirar", 120)]  # simulan al usuario

for tipo, importe in operaciones:              # repetición
    if tipo == "retirar":                      # selección
        if importe <= saldo:
            saldo -= importe
            print(f"Retiras {importe} € -> saldo {saldo} €")
        else:
            print(f"Operación rechazada ({importe} €): saldo insuficiente ({saldo} €)")
    elif tipo == "ingresar":
        saldo += importe
        print(f"Ingresas {importe} € -> saldo {saldo} €")

print(f"Saldo final: {saldo} €")      # secuencial
````

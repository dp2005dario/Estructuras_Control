## Ejercicio 4 · Criterio **d)** · ⭐⭐

Escribe `maximo_y_minimo(lista)` que devuelva la tupla `(maximo, minimo)` **sin usar `max()` ni `min()`**: recorre la lista con un bucle y actualiza dos variables con condicionales. Si la lista está vacía, devuelve `None`. *(Pista: inicializa ambas variables con el primer elemento.)*

````python
def maximo_y_minimo(lista):
    # Si la lista está vacía, devolvemos None tal como pide el enunciado
    if not lista:
        return None
    
    # Inicializamos el máximo y el mínimo con el primer elemento
    maximo = lista[0]
    minimo = lista[0]
    
    # Recorremos la lista con un bucle
    for elemento in lista:
        if elemento > maximo:
            maximo = elemento
        if elemento < minimo:
            minimo = elemento
            
    # Devolvemos el resultado en forma de tupla (máximo, mínimo)
    return (maximo, minimo)
    pass
````

````python
# Comprobación (NO modificar)
assert maximo_y_minimo([4, 9, -2, 7]) == (9, -2)
assert maximo_y_minimo([-5, -3, -8]) == (-3, -8), "Cuidado con las listas de negativos"
assert maximo_y_minimo([6]) == (6, 6)
assert maximo_y_minimo([]) is None
print("✔ Ejercicio 4 superado")
````

## Ejercicio 5 · Criterio **f)** · ⭐

Escribe `letra_nota(nota)`:

| Nota | Resultado |
|---|---|
| menor que 0 o mayor que 10 | `None` |
| de 0 a menos de 5 | `"F"` |
| de 5 a menos de 6 | `"D"` |
| de 6 a menos de 7 | `"C"` |
| de 7 a menos de 9 | `"B"` |
| de 9 a 10 | `"A"` |

````python
def letra_nota(nota):
    # Si la nota es menor que 0 o mayor que 10, devuelve None
    if nota < 0 or nota > 10:
        return None
    elif nota < 5:
        return "F"
    elif nota < 6:
        return "D"
    elif nota < 7:
        return "C"
    elif nota < 9:
        return "B"
    else:  # Desde 9 hasta 10 (incluidos)
        return "A"
    pass
````

````python
# Comprobación (NO modificar)
casos = {-1: None, 0: "F", 4.99: "F", 5: "D", 5.99: "D", 6: "C", 6.5: "C", 7: "B", 8.99: "B", 9: "A", 10: "A", 10.1: None}
for nota, esperado in casos.items():
    assert letra_nota(nota) == esperado, f"Falla con {nota}: esperado {esperado}"
print("✔ Ejercicio 5 superado")
````

## Ejercicio 6 · Criterio **i)** · ⭐⭐

Escribe estas dos funciones con sentencias condicionales:

1. `es_bisiesto(anio)`: `True` si el año es divisible por 4 **y no** por 100, **o** es divisible por 400.
2. `dias_del_mes(mes, anio)`: días del mes (1-12) teniendo en cuenta los bisiestos (febrero tiene 29 días si el año es bisiesto). Si el mes no es válido, devuelve `None`. *(Usa `es_bisiesto`.)*

````python
def es_bisiesto(anio):
    if anio %4 ==0 and anio %100!=0 or anio %400==0:
        return True

def dias_del_mes(mes, anio):
    if mes in [1, 3, 5, 7, 8, 10, 12]:
        return 31
    elif mes in [4, 6, 9, 11]:
        return 30
    elif mes == 2:
        if es_bisiesto(anio):
            return 29
        else:
            return 28
    else:
        return None
````

````python
# Comprobación (NO modificar)
assert [es_bisiesto(a) for a in (2024, 2023, 1900, 2000, 2100)] == [True, False, False, True, False]
assert dias_del_mes(2, 2024) == 29 and dias_del_mes(2, 2023) == 28
assert [dias_del_mes(m, 2023) for m in (1, 4, 6, 9, 11, 12)] == [31, 30, 30, 30, 30, 31]
assert dias_del_mes(0, 2023) is None and dias_del_mes(13, 2023) is None
print("✔ Ejercicio 6 superado")
````

## Ejercicio 8 · Criterio **j)** · ⭐⭐⭐

Escribe `elegir_transporte(distancia_km, llueve, tiene_bici)` **con condicionales anidadas** (un `if` dentro de otro) que devuelva:

- Si `distancia_km <= 1`: `"andando con paraguas"` si llueve, o `"andando"` si no.
- Si `distancia_km <= 5`: si tiene bici → `"autobús"` si llueve o `"bicicleta"` si no; si no tiene bici → `"autobús"`.
- Si la distancia es mayor que 5: `"metro"`.

````python
def elegir_transporte(distancia_km, llueve, tiene_bici):
    if distancia_km <= 1:
        if llueve:
            return "andando con paraguas"
        else:
            return "andando"
            
    elif distancia_km <= 5:
        if tiene_bici:
            if llueve:
                return "autobús"
            else:
                return "bicicleta"
        else:
            return "autobús"
            
    else:  # Si la distancia es mayor que 5
        return "metro"
````

````python
# Comprobación (NO modificar)
casos = [
    ((0.5, False, True), "andando"), ((1, True, False), "andando con paraguas"),
    ((3, False, True), "bicicleta"), ((3, True, True), "autobús"), ((5, False, False), "autobús"),
    ((5.1, False, True), "metro"), ((20, True, False), "metro"),
]
for args, esperado in casos:
    assert elegir_transporte(*args) == esperado, f"Falla con {args}: esperado {esperado}"
print("✔ Ejercicio 8 superado (el profesorado comprueba que hay anidación)")
````

## Ejercicio 9 · Criterio **h)** · ⭐

Escribe estas funciones con **una sola línea** (operador ternario u operador `or`), sin `abs()`:

1. `valor_absoluto(n)`
2. `precio_final(precio, es_socio)`: el 10 % de descuento si `es_socio`, o el precio sin cambios.
3. `saludo(nombre)`: `"Hola, <nombre>"`, o `"Hola, anónimo"` si el nombre es una cadena vacía (usa `or`).

````python
def valor_absoluto(n):
    return n if n >= 0 else -n
    pass

def precio_final(precio, es_socio):
    return precio * 0.9 if es_socio else precio
    pass

def saludo(nombre):
    return f"Hola, {nombre}" if nombre else "Hola, anónimo"   
    pass
````

````python
# Comprobación (NO modificar)
assert valor_absoluto(-7) == 7 and valor_absoluto(5) == 5 and valor_absoluto(0) == 0
assert precio_final(100, True) == 90.0 and precio_final(100, False) == 100
assert saludo("Ana") == "Hola, Ana" and saludo("") == "Hola, anónimo"
print("✔ Ejercicio 9 superado (el profesorado comprueba el uso del ternario)")
````

## 1️⃣ Conceptos clave: ventana, widgets y eventos

Un programa con interfaz gráfica (GUI) en Tkinter se compone de:

| Concepto | Qué es | Ejemplo |
|---|---|---|
| **Ventana** (*window*) | El contenedor principal del programa | La ventana con la barra de título y la X |
| **Widget** | Cada elemento visual dentro de la ventana | Botón, caja de texto, etiqueta… |
| **Evento** | Algo que hace el usuario | Pulsar un botón, escribir, mover el ratón |
| **Callback** | Función que se ejecuta cuando ocurre un evento | La función que convierte km a millas |

Los programas que has hecho hasta ahora se ejecutan **de arriba abajo y terminan**. Un programa con GUI es distinto: **se queda esperando** a que el usuario haga algo y reacciona. Eso se llama **programación orientada a eventos**.

## 4️⃣ Añadir un widget: el botón

Cada widget se crea **llamando a su clase**. El **primer argumento** siempre es el **padre**: el contenedor donde vivirá (por ahora, la ventana).

```python
boton = ttk.Button(window, text="Ejecutar")
```

Pero... ¡crear el widget no lo muestra! Hay que **colocarlo** con un *gestor de geometría* (`pack` o `grid`). Comprobémoslo:

````python
window = tk.Tk()
window.title("Botón sin colocar")
window.geometry("300x100")

b1 = ttk.Button(window, text="Ejecutar")   # creado... pero NO colocado

traer_al_frente(window)
window.mainloop()     # 👀 ¿Ves el botón? No: falta colocarlo.
````

### Colocar con `pack()` y con `grid()`

- **`pack()`** apila los widgets uno tras otro. Es lo más rápido para cosas simples.
- **`grid()`** los coloca en una **cuadrícula de filas y columnas**. Da mucho más control y es el que usaremos.

Veamos primero `pack()`:

````python
window = tk.Tk()
window.title("Botón con pack")
window.geometry("300x100")

b1 = ttk.Button(window, text="Ejecutar")
b1.pack()                                  # ✅ ahora sí aparece

traer_al_frente(window)
window.mainloop()
````

## 5️⃣ Organizar con `grid()`

Imagina la ventana como una **hoja cuadriculada**. Cada widget ocupa una celda indicada por `row` (fila) y `column` (columna). **Se empieza a contar desde 0**.

```
            column=0      column=1      column=2
row=0   [ Botón (0,0) ] [ Entrada (0,1) ] [ Texto (0,2) ]
row=1   [     ...     ] [     ...      ] [     ...     ]
```

Opciones muy útiles de `grid()`:

| Opción | Qué hace |
|---|---|
| `row`, `column` | Posición en la cuadrícula (desde 0) |
| `rowspan`, `columnspan` | Cuántas filas / columnas ocupa el widget |
| `padx`, `pady` | Margen exterior horizontal / vertical (en píxeles) |
| `sticky` | Hacia qué lados "se pega" el widget: `"w"` (izquierda), `"e"` (derecha), `"ew"` (estirar horizontalmente)… |

> ⛔ **Regla de oro:** dentro de un mismo contenedor, **no mezcles `pack()` y `grid()`**. Elige uno. Tkinter puede quedarse bloqueado o dar un error.

````python
window = tk.Tk()
window.title("Botón con grid")

b1 = ttk.Button(window, text="Ejecutar")
b1.grid(row=0, column=0, padx=10, pady=10)

traer_al_frente(window)
window.mainloop()
````

## 7️⃣ Dar vida al botón: funciones *callback*

Para que un botón haga algo se usa el parámetro **`command`**, al que le pasamos **una función**. Cuando el usuario pulsa el botón, Tkinter *llama de vuelta* (*call back*) a esa función.

### 🔑 La regla más importante: **sin paréntesis**

```python
command=mi_funcion      # ✅ pasas la función (Tkinter la llamará cuando haya clic)
command=mi_funcion()    # ❌ la ejecutas AHORA y pasas su resultado (None)
```

Comprobémoslo con un experimento. Primero el **error**:

````python
def saludar():
    print("¡Función ejecutada!")

window = tk.Tk()
window.title("Error de los paréntesis")

# ❌ Con paréntesis: "¡Función ejecutada!" aparece YA, al crear el botón,
#    y después el botón no hace nada al pulsarlo.
b1 = ttk.Button(window, text="Pulsa", command=saludar())
b1.grid(row=0, column=0, padx=20, pady=20)

traer_al_frente(window)
window.mainloop()
````

Ahora la forma **correcta**. Pulsa varias veces y mira la salida debajo de la celda:

````python
def saludar():
    print("¡Función ejecutada!")

window = tk.Tk()
window.title("Callback correcto")

# ✅ Sin paréntesis: pasamos la función, no su resultado
b1 = ttk.Button(window, text="Pulsa", command=saludar)
b1.grid(row=0, column=0, padx=20, pady=20)

traer_al_frente(window)
window.mainloop()
````

## 🔟 Procesar el valor: el error `TypeError` y la conversión a `float`

Ya copiamos el texto de la entrada al `Text`. Ahora queremos **calcular** algo con él. Si intentamos multiplicar lo que devuelve `get()` por un número decimal... ¡error! Veámoslo **sin ventana**, en Python puro, para entenderlo bien:

````python
valor = "10"          # esto es lo que devuelve get(): un STRING, no un número

try:
    resultado = valor * 1.6
except TypeError as error:
    print("TypeError:", error)

# Curiosidad: un str SÍ se puede multiplicar por un entero (repite el texto)
print("me" * 2)
````

**¿Por qué ocurre?** Un `str` es una *secuencia* de caracteres. Python solo sabe multiplicar una secuencia por un **entero** (para repetirla). Con un decimal (`1.6`) no tiene sentido y lanza `TypeError`.

**Solución:** convertir el texto a número con `float()`:

````python
valor = "10"
km = float(valor)       # "10" -> 10.0
print(km * 1.6)
````

## 1️⃣1️⃣ 🛠️ ¡Verifica la fórmula de conversión!

Pensemos la fórmula para realizar la conversión entre unidades:

- 1 **milla** = 1,609344 **kilómetros** (una milla es más larga que un kilómetro).
- Por tanto, **1 km ≈ 0,621371 millas**, y `millas = km / 1.609344`.

El valor exacto es **exactamente 1,609344** (definido por convenio internacional).

Comprobación de sentido común: 10 km deberían ser *menos* de 10 millas (≈ 6,21), no más.

````python
# Constante con nombre (en MAYÚSCULAS por convención): evita "números mágicos"
KM_POR_MILLA = 1.609344

def km_a_millas(km):
    """Convierte kilómetros a millas."""
    return km / KM_POR_MILLA

print("Fórmula correcta:               10 km ->", round(km_a_millas(10), 2), "millas")
````

### 🧩 Buena práctica: separar la **lógica** de la **interfaz**

Fíjate que `km_a_millas` **no sabe nada de Tkinter**: recibe un número y devuelve un número. Esto tiene ventajas enormes:

- Se puede **probar** sin abrir ninguna ventana.
- Se puede **reutilizar** en otros programas.
- Es la misma idea del proyecto final del curso: *"frontend"* (la interfaz) y *"backend"* (el código que hace el trabajo).

Probémosla con `assert`: si la afirmación es falsa, Python lanza un error; si es cierta, no dice nada.

````python
assert abs(km_a_millas(1.609344) - 1) < 1e-9     # 1.609344 km son exactamente 1 milla
assert abs(km_a_millas(0)) == 0
assert round(km_a_millas(10), 2) == 6.21
print("✅ Todas las pruebas pasan")
````

## 1️⃣2️⃣ Que nada rompa tu programa: validar la entrada

¿Qué pasa si el usuario escribe `hola`, deja la caja vacía o escribe `10,5` (con coma, como se hace en España)? `float("hola")` lanza `ValueError` y `float("10,5")` también.

Para controlarlo:

1. Sustituimos la coma por punto (`replace(",", ".")`).
2. Quitamos espacios sobrantes (`strip()`).
3. Intentamos convertir dentro de `try/except ValueError`.
4. Si falla, avisamos al usuario con una ventana de mensaje (`messagebox.showerror`).

Probemos esta lógica antes de meterla en la GUI:

````python
def leer_numero(texto):
    """Convierte un texto a float aceptando coma o punto decimal.
    Devuelve None si no es un número válido."""
    try:
        return float(texto.strip().replace(",", "."))
    except ValueError:
        return None

for prueba in ["10", "10,5", "  3.2 ", "hola", ""]:
    print(repr(prueba), "->", leer_numero(prueba))
````

## 1️⃣3️⃣ 🏁 El programa completo

Juntamos todas las piezas. Fíjate en la estructura, que es la que seguirás en programas más grandes:

1. **Imports y constantes**
2. **Funciones de lógica** (sin Tkinter)
3. **Callback** (conecta la interfaz con la lógica)
4. **Interfaz**: ventana y widgets
5. **`mainloop()`**

````python
import tkinter as tk
from tkinter import ttk, messagebox

# ---------- 1. Constantes ----------
KM_POR_MILLA = 1.609344


# ---------- 2. Lógica (no depende de Tkinter) ----------
def km_a_millas(km):
    return km / KM_POR_MILLA


def leer_numero(texto):
    try:
        return float(texto.strip().replace(",", "."))
    except ValueError:
        return None


# ---------- 3. Callback ----------
def convertir(event=None):       # event=None: sirve para el botón Y para la tecla Enter
    km = leer_numero(entrada.get())

    if km is None:
        messagebox.showerror("Dato no válido", "Escribe un número, por ejemplo 10 o 12,5")
        return

    millas = km_a_millas(km)
    salida.delete("1.0", tk.END)                 # limpiar el resultado anterior
    salida.insert(tk.END, f"{millas:.2f}")      # escribir el nuevo (2 decimales)


# ---------- 4. Interfaz ----------
window = tk.Tk()
window.title("Conversor km → millas")
window.resizable(False, False)

etiqueta = ttk.Label(window, text="Kilómetros:")
etiqueta.grid(row=0, column=0, padx=(10, 5), pady=15)

entrada = ttk.Entry(window, width=12)
entrada.grid(row=0, column=1, padx=5, pady=15)

boton = ttk.Button(window, text="Convertir a millas", command=convertir)
boton.grid(row=0, column=2, padx=5, pady=15)

salida = tk.Text(window, height=1, width=14)
salida.grid(row=0, column=3, padx=(5, 10), pady=15)

entrada.focus()                       # el cursor empieza en la caja
window.bind("<Return>", convertir)    # Enter = pulsar el botón

# ---------- 5. Bucle de eventos ----------
traer_al_frente(window)
window.mainloop()
````

| `1,609344` | `1.00` |
| `0` | `0.00` |
| `hola` | Ventana de error |
| *(vacío)* | Ventana de error |

### 🧠 ¿Qué has aprendido en esta versión?
- `command=convertir` (sin paréntesis) → el botón llama a la función.
- `window.bind("<Return>", convertir)` → los **eventos de teclado** también llaman a funciones. Por eso `convertir` recibe un parámetro `event` (Tkinter se lo envía). Con `event=None` por defecto, vale para ambos usos.
- Separar `km_a_millas` y `leer_numero` de la interfaz hace el código más limpio y comprobable.

## 1️⃣4️⃣ Guardarlo como programa independiente (`.py`)

Los programas con ventana se suelen usar **fuera** del cuaderno. Esta celda usa la "magia" `%%writefile` para crear el archivo `conversor_km_millas.py` en la carpeta del cuaderno.

La única novedad es `if __name__ == "__main__":`, que significa *"ejecuta esto solo si lanzo este archivo directamente (no si lo importo desde otro)"*. Es la forma estándar de estructurar un script.

````python
%%writefile conversor_km_millas.py
"""Conversor de kilómetros a millas con Tkinter."""
import tkinter as tk
from tkinter import ttk, messagebox

KM_POR_MILLA = 1.609344


def km_a_millas(km):
    return km / KM_POR_MILLA


def leer_numero(texto):
    try:
        return float(texto.strip().replace(",", "."))
    except ValueError:
        return None


def crear_interfaz():
    window = tk.Tk()
    window.title("Conversor km → millas")
    window.resizable(False, False)

    def convertir(event=None):
        km = leer_numero(entrada.get())
        if km is None:
            messagebox.showerror("Dato no válido", "Escribe un número, por ejemplo 10 o 12,5")
            return
        salida.delete("1.0", tk.END)
        salida.insert(tk.END, f"{km_a_millas(km):.2f}")

    ttk.Label(window, text="Kilómetros:").grid(row=0, column=0, padx=(10, 5), pady=15)

    entrada = ttk.Entry(window, width=12)
    entrada.grid(row=0, column=1, padx=5, pady=15)

    ttk.Button(window, text="Convertir a millas", command=convertir).grid(row=0, column=2, padx=5, pady=15)

    salida = tk.Text(window, height=1, width=14)
    salida.grid(row=0, column=3, padx=(5, 10), pady=15)

    entrada.focus()
    window.bind("<Return>", convertir)
    return window


if __name__ == "__main__":
    crear_interfaz().mainloop()
````

Ahora ábrelo **desde una terminal** (en la carpeta del cuaderno):

```bash
python conversor_km_millas.py        # en algunos sistemas: python3 ...
```

> 💡 En la versión del archivo hemos metido toda la interfaz dentro de una función `crear_interfaz()`. Es un paso más hacia un código ordenado: al importar el módulo no se abre nada, y puedes crear la ventana cuando quieras.

### Errores comunes y su causa
| Síntoma | Causa probable |
|---|---|
| `AttributeError: ... no attribute 'tk'` | Escribiste `tk()` en vez de `tk.Tk()` |
| La ventana se abre y cierra al instante | Falta `window.mainloop()` |
| El widget no aparece | Falta `.grid()` o `.pack()` |
| La función se ejecuta al abrir, y el botón no hace nada | `command=funcion()` con paréntesis |
| `TypeError: can't multiply sequence by non-int of type 'float'` | `get()` devuelve `str`; falta `float()` |
| `ValueError: could not convert string to float` | El usuario escribió algo que no es un número; falta validar |
| `TclError` al mezclar `pack` y `grid` | Usaste ambos en el mismo contenedor |

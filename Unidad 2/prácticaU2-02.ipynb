# 1. Práctica U2 - 02 - Calculadora de precio final

## 1.1 Objetivo
El objetivo de este ejercicio es desarrollar una aplicación en Python que calcule el **precio final de un producto** a partir de su precio, un descuento y el tipo de IVA aplicable. Para ello, deberás crear una interfaz gráfica utilizando Tkinter (TTK preferentemente) y organizar el código en una clase (no es obligatorio, pero puntúa más). El programa debe permitir al usuario elegir el tipo de IVA, introducir el precio y el descuento, indicar si el precio ya incluye IVA, y mostrar en pantalla el desglose del cálculo.

## 1.2 Requisitos
### 1.2.1. Interfaz gráfica
   - La ventana principal debe tener un título adecuado y un tamaño de 360x400 píxeles.
   - Añade un widget de tipo Sizegrip (TTK) en la esquina inferior derecha con 3 píxeles de margen en todas las direcciones.
   - Añade un marco TTK a la ventana principal con un margen superior de 10 píxeles y un margen lateral de 10 píxeles. Este marco será el de **datos** y contendrá:
        - Un ComboBox TTK de **solo lectura** para seleccionar el tipo de IVA (ver apartado 1.3).
        - Un campo de entrada TTK Entry para el precio en euros.
        - Un Spinbox TTK para el porcentaje de descuento, con valores de 0 a 100 en incrementos de 5 y valor inicial 0.
        - Una etiqueta TTK junto a cada uno de los tres campos anteriores indicando su finalidad. Los campos deben ocupar todo el ancho disponible y guardar márgenes verticales adecuados.
        - Un Checkbutton TTK con el texto "El precio ya incluye IVA", desmarcado por defecto.
   - Añade un segundo marco TTK a la ventana principal con un margen superior de 10 píxeles. Este marco contendrá:
        - Dos botones TTK, "Calcular" y "Limpiar", colocados **uno al lado del otro** y repartiéndose a partes iguales el ancho del marco.
        - Una línea de separación horizontal (TTK) que rellene el ancho del marco, con algo de margen.
        - Una etiqueta TTK donde se mostrará el resultado del cálculo o un mensaje de error. Agrega margen.

### 1.2.2. Funcionalidad
   - Al pulsar "Calcular" se validarán los datos y, si son correctos, se mostrará en la etiqueta el desglose del cálculo en **cinco líneas**: base imponible, descuento aplicado, base con descuento, IVA y total.
   - Todos los importes se mostrarán con **dos decimales** y el símbolo del euro, y deben quedar alineados en columna. Pista: una etiqueta puede usar una fuente de ancho fijo con `font="TkFixedFont"`.
   - El cálculo se realizará así:
        - Si el precio **no** incluye IVA, la base imponible es el precio introducido.
        - Si el precio **sí** incluye IVA, la base imponible es `precio / (1 + IVA)`, donde IVA es el tipo seleccionado expresado en tanto por uno (por ejemplo, 0.21).
        - Importe del descuento = `base × descuento / 100`.
        - Base con descuento = base − importe del descuento.
        - Importe del IVA = base con descuento × IVA.
        - Total = base con descuento + importe del IVA.
   - Al pulsar "Limpiar" se devolverá la aplicación a su estado inicial: ComboBox sin selección, precio vacío, descuento a 0, Checkbutton desmarcado y etiqueta de resultado vacía.
   - Los mensajes de error se mostrarán en color rojo y el resultado correcto en color negro.

### 1.2.3. Clases
   - El código preferentemente estará organizado en una clase llamada `Ventana`. Esta clase debe manejar tanto la creación de la interfaz gráfica como la lógica del cálculo.
   - La clase `Ventana` debe contener métodos para:
        - Inicializar la ventana y los widgets (ComboBox, campos de entrada, botones, etc.).
        - Realizar el cálculo a partir de los datos introducidos y mostrar el resultado.
        - Restablecer la aplicación a su estado inicial.

## 1.5 Validaciones
   - Si no se ha seleccionado un tipo de IVA, el programa debe mostrar un mensaje de error. Emplea, por ejemplo, `mi_etiqueta.config(text="No se ha seleccionado el tipo de IVA", foreground="red")`.
   - Si el precio no es un número, el programa debe mostrar un mensaje de error. Usa `try-except`. Se admitirá tanto el punto como la coma como separador decimal (`12.5` y `12,5` son válidos).
   - Si el precio es negativo, el programa debe mostrar un mensaje de error.
   - Si el descuento no es un número, o no está entre 0 y 100 (ambos incluidos), el programa debe mostrar un mensaje de error. Recuerda que un Spinbox permite escribir texto libre.
   - Ten en cuenta que `float()` acepta cadenas como `"nan"` o `"inf"`: estos valores **no** deben considerarse válidos.

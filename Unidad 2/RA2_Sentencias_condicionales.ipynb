## 1. Concepto de sentencia condicional (criterio a)

Una **sentencia condicional** (o de **selección**) permite que el programa **tome decisiones**: ejecuta un bloque de código **solo si** una condición es verdadera. Es la forma de que un programa reaccione de manera distinta ante datos distintos.

> Analogía: *"Si llueve, cojo el paraguas; si no, me pongo las gafas de sol."*

Python dispone de:

| Sentencia | Uso |
|---|---|
| `if` | Ejecuta un bloque si la condición es cierta |
| `if ... else` | Elige entre dos caminos |
| `if ... elif ... else` | Elige entre varios caminos |
| `if` dentro de `if` | Condicionales **anidadas** |
| `x if cond else y` | Expresión condicional (forma **compacta**) |
| `match ... case` | Selección por patrones (Python ≥ 3.10) |

````python
llueve = True

if llueve:
    print("Cojo el paraguas ☔")
else:
    print("Me pongo las gafas de sol 😎")
````

## 2. Partes de una sentencia condicional (criterio b)

```python
if edad >= 18:            # ① palabra clave  ② condición  ③ dos puntos
    print("Mayor")        # ④ bloque (sangrado)
elif edad >= 14:          # ⑤ rama alternativa (opcional)
    print("Adolescente")
else:                     # ⑥ rama por defecto (opcional, sin condición)
    print("Niño")
```

| Parte | Descripción |
|---|---|
| ① Palabra clave | `if`, `elif`, `else` |
| ② Condición | Expresión que se evalúa como `True`/`False` (`else` **no** lleva condición) |
| ③ Dos puntos `:` | Obligatorios al final de cada cabecera |
| ④ Bloque | Instrucciones sangradas que se ejecutan si se cumple |
| ⑤ `elif` | 0 o más; se evalúan en orden hasta que una es verdadera |
| ⑥ `else` | 0 o 1; siempre la última |

**Reglas:** solo se ejecuta **una** rama; en cuanto una condición es verdadera, las demás se ignoran.

````python
# Identifica las partes en este ejemplo
edad = 16

if edad >= 18:
    categoria = "Mayor"
elif edad >= 14:
    categoria = "Adolescente"
else:
    categoria = "Niño"

print(categoria)
````

## 3. El sangrado en las sentencias condicionales (criterio c)

El **sangrado** indica qué instrucciones pertenecen a cada rama. Usa **4 espacios** por nivel.

````python
x = 10

if x > 5:
    print("x es mayor que 5")        # pertenece al if
    print("Sigue en el if")          # pertenece al if
else:
    print("x es menor o igual a 5")  # pertenece al else
print("Esto se ejecuta siempre")     # fuera de la condicional
````

### 3.1 Un sangrado erróneo cambia el resultado (¡sin error!)

````python
numero = 3

# Versión A: ambas líneas dentro del if
if numero > 5:
    print("Grande")
    print("Fin del análisis")

print("-----")

# Versión B: la segunda línea queda fuera
if numero > 5:
    print("Grande")
print("Fin del análisis")
````

## 4. Ejecución condicional y control de variables (criterio d)

Las condicionales controlan **qué instrucciones se ejecutan** y, con ello, **cómo cambian las variables**. Usos habituales:

1. **Modificar una variable solo si se cumple algo** (acumular, contar).
2. **Validar datos antes de usarlos** (evitar errores como la división por cero).
3. **Banderas (*flags*)**: variables booleanas que recuerdan que algo ha ocurrido.
4. **Asignar un valor por defecto.**

````python
# 1) Modificar variables condicionalmente: contar pares e impares
numeros = [4, 7, 10, 3, 8, 5, 1]
pares = 0
impares = 0

for n in numeros:
    if n % 2 == 0:
        pares += 1
    else:
        impares += 1

print(f"Pares: {pares} | Impares: {impares}")
````

````python
# 2) Validar antes de operar: evitar la división por cero
dividendo, divisor = 10, 0

if divisor != 0:
    resultado = dividendo / divisor
else:
    resultado = None
    print("No se puede dividir entre cero")

print("Resultado:", resultado)
````

````python
# 3) Bandera (flag): ¿hay algún suspenso?
notas = [6, 8, 4, 9]
hay_suspensos = False

for nota in notas:
    if nota < 5:
        hay_suspensos = True

if hay_suspensos:
    print("Hay al menos un suspenso")
else:
    print("Todos aprobados")
````

````python
# 4) Valor por defecto
nombre = ""          # imagina que el usuario no escribió nada
if not nombre:
    nombre = "Anónimo"
print("Hola,", nombre)
````

````python
# Control de variables: ¡cuidado con las variables que solo se definen en una rama!
try:
    if False:
        mensaje = "definida"
    print(mensaje)
except NameError as e:
    print("NameError:", e)

# Solución: inicializar la variable antes del if
mensaje = "valor por defecto"
if False:
    mensaje = "definida"
print(mensaje)
````

### ✏️ Actividad 4
Dada la lista de temperaturas, calcula con condicionales: el **máximo**, el **mínimo** (sin usar `max`/`min`) y cuántos días hicieron más de 30 °C.

````python
temperaturas = [22, 31, 35, 28, 19, 33, 30]

# Es una buena práctica inicializar el máximo y mínimo con el primer elemento de la lista
maximo = temperaturas[0]
minimo = temperaturas[0]
dias_calurosos = 0

for temp in temperaturas:
    # Comprobamos si hay un nuevo máximo
    if temp > maximo:
        maximo = temp
        
    # Comprobamos si hay un nuevo mínimo
    if temp < minimo:
        minimo = temp
        
    # Contamos los días que hicieron estrictamente más de 30 ºC
    if temp > 30:
        dias_calurosos += 1

print(f"Temperatura máxima: {maximo} ºC")
print(f"Temperatura mínima: {minimo} ºC")
print(f"Días con más de 30 ºC: {dias_calurosos}")
````

## 5. Funcionamiento de las sentencias condicionales (criterio e)

### 5.1 Evaluación en orden y cortocircuito

- Las ramas se evalúan **de arriba abajo**; se ejecuta la primera cuya condición sea verdadera.
- `and` y `or` usan **cortocircuito**: si el resultado ya se conoce, no se evalúa el resto.

````python
def hablar(texto, valor):
    print("  evaluando", texto)
    return valor

print("Caso and (primera falsa):")
if hablar("A", False) and hablar("B", True):
    pass

print("Caso or (primera verdadera):")
if hablar("A", True) or hablar("B", False):
    pass
````

````python
# El ORDEN de las condiciones importa
nota = 9

print("Orden incorrecto:")
if nota >= 5:
    print("  Aprobado")        # <- se ejecuta y se ignoran las demás
elif nota >= 9:
    print("  Sobresaliente")   # nunca llega aquí

print("Orden correcto:")
if nota >= 9:
    print("  Sobresaliente")
elif nota >= 5:
    print("  Aprobado")
````

## 6. Aplicación correcta de las sentencias condicionales (criterio f)

Buenas prácticas:

- Elige `if/elif/else` cuando las opciones son **excluyentes**; usa varios `if` cuando son **independientes**.
- Evita condiciones redundantes (`if x == True` → `if x`).
- Cuida los **límites** (`<` vs `<=`).
- Usa `in` para comparar con varios valores.
- Evita repetir código: saca lo común fuera de las ramas.

````python
# Opciones excluyentes -> if/elif/else
dia = "sábado"
if dia in ("sábado", "domingo"):      # 'in' evita (dia == "sábado" or dia == "domingo")
    print("Fin de semana")
else:
    print("Día laborable")
````

````python
# Opciones independientes -> varios if
carrito = 120
cliente_vip = True
envio_gratis = False

if carrito > 100:
    print("Descuento del 5 % por compra superior a 100 €")
if cliente_vip:
    print("Descuento VIP adicional")
if carrito > 50 and cliente_vip:
    envio_gratis = True
print("Envío gratis:", envio_gratis)
````

````python
# Mal estilo vs. buen estilo
activo = True

# Mal estilo
if activo == True:
    print("activo (mal estilo)")

# Buen estilo
if activo:
    print("activo (buen estilo)")

# Comparaciones encadenadas: más legibles
edad = 25
if 18 <= edad < 65:
    print("Edad laboral")
````

````python
# Ejemplo de aplicación: año bisiesto
anio = 2100

if (anio % 4 == 0 and anio % 100 != 0) or anio % 400 == 0:
    print(f"{anio} es bisiesto")
else:
    print(f"{anio} no es bisiesto")
````

## 7. Anidaciones (criterio g)

Una condicional **anidada** es una condicional dentro de otra. La interior solo se evalúa si se cumple la exterior. Cada nivel añade **4 espacios** de sangrado.

```python
if condicion_exterior:
    if condicion_interior:
        ...
    else:
        ...
else:
    ...
```

### Diagrama de flujo

```mermaid
flowchart TD
    A{¿Tiene tarjeta?} -- Sí --> B{¿Pin correcto?}
    A -- No --> C[Rechazar]
    B -- Sí --> D[Permitir acceso]
    B -- No --> E[Pedir de nuevo]
```

````python
tiene_tarjeta = True
pin_correcto = False

if tiene_tarjeta:
    if pin_correcto:
        print("Acceso permitido")
    else:
        print("PIN incorrecto")
else:
    print("Inserte una tarjeta")
````

### 7.1 ¿Anidar o combinar con `and`?

Si no hay un `else` distinto para cada nivel, se puede **aplanar** con `and` (más legible). Anidar es mejor cuando cada nivel tiene su propia respuesta, o cuando la condición interior solo tiene sentido si la exterior se cumple (por ejemplo, comprobar que un divisor no es cero antes de usarlo).

````python
edad, carnet = 20, True

# Anidada
if edad >= 18:
    if carnet:
        print("Puede conducir")

# Aplanada (equivalente)
if edad >= 18 and carnet:
    print("Puede conducir")
````

````python
# Anidación con elif: clasificación de un triángulo por sus lados
a, b, c = 5, 5, 8

if a + b > c and a + c > b and b + c > a:        # ¿existe el triángulo?
    if a == b == c:
        tipo = "equilátero"
    elif a == b or a == c or b == c:
        tipo = "isósceles"
    else:
        tipo = "escaleno"
    print(f"Triángulo {tipo}")
else:
    print("No forma un triángulo")
````

> ⚠️ **Consejo:** más de 3 niveles de anidación suelen indicar que hay que simplificar (combinando condiciones, usando *cláusulas de guarda* o funciones).

````python
# Cláusulas de guarda: alternativa a la anidación profunda (dentro de una función)
def clasificar_nota(nota):
    if not isinstance(nota, (int, float)):
        return "Valor no numérico"
    if nota < 0 or nota > 10:
        return "Nota fuera de rango"
    if nota >= 5:
        return "Aprobado"
    return "Suspenso"

for valor in (7, 3, 11, "a"):
    print(valor, "->", clasificar_nota(valor))
````

## 8. Estructuras compactas (criterio h)

### 8.1 Expresión condicional (operador ternario)

```python
valor_si_cierto if condicion else valor_si_falso
```

````python
edad = 20
estado = "mayor" if edad >= 18 else "menor"
print(estado)

# Equivalente en forma larga
if edad >= 18:
    estado = "mayor"
else:
    estado = "menor"
````

````python
# Dentro de f-strings y llamadas a funciones
n = 5
print(f"{n} es {'par' if n % 2 == 0 else 'impar'}")

# Ternarios encadenados (usar con moderación: se leen mal)
nota = 6
txt = "Suspenso" if nota < 5 else "Aprobado" if nota < 9 else "Sobresaliente"
print(txt)
````

### 8.4 `match ... case` (Python ≥ 3.10)

Alternativa a las cadenas largas de `elif` que comparan una variable con varios valores.

````python
import sys
print("Versión de Python:", sys.version.split()[0])

if sys.version_info >= (3, 10):
    exec('''
comando = "parar"
match comando:
    case "arrancar":
        print("Motor en marcha")
    case "parar" | "detener":
        print("Motor detenido")
    case _:
        print("Comando desconocido")
''')
else:
    print("match/case requiere Python 3.10 o superior")
````
